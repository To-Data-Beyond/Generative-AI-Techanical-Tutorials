# Write Python Clean Code Using These 3 Principles

Companion to the [complete article](https://todatabeyond.com/blog/write-python-clean-code-using-these-3-principles).

Run all cells in order with Python 3.10 or newer. Standard library only; no package installation, data downloads, API keys, network calls, database or GPU. Every one of the thirteen original gist examples is retained, including both recommended and nonrecommended versions.

Maintenance repairs smart quotes, missing indentation and uppercase `Pass`, and supplies missing variable/function context. The `pass` fetch functions are deliberately naming-only skeletons, not fake network services. The comments example uses explicitly labelled in-memory `User`/`Response` stand-ins, not Django or real HTTP; regression tests check only this local demonstration. The two dice-naming examples retain the source’s inclusive 0–4 range; they illustrate naming, not a physical four-sided die. Intentionally poor indentation remains compilable Python for the style comparison. No execution counts or outputs are committed.


## 1. Descriptive variable names


```python
# Not recommended
# The au variable is the number of active users
au = 105

# Recommended 
total_active_users = 105

print(au, total_active_users)
```

## 2. Readable sequence names


```python
# Not recommended
c = ["UK", "USA", "UAE"]
for x in c:
    print(x)

# Recommended
cities_list = ["UK", "USA", "UAE"]
for city in cities_list:
    print(city)
```

## 3. Consistent vocabulary


```python
# Not recommended
client_first_name = 'John'
customer_last_name = 'Doe'

# Recommended
client_first_name = 'John'
client_last_name = 'Doe'

# Another example:

# bad code
def fetch_clients(response, variable):
    # do something
    pass

def fetch_posts(res, var):
    # do something
    pass

# Recommended
def fetch_clients(response, variable):
    # do something
    pass

def fetch_posts(response, variable):
    # do something
    pass

print(client_first_name, client_last_name)
print(fetch_clients(None, None), fetch_posts(None, None))
```

## 4. Named constants


```python
import random

# Not recommended
def roll_dice():
    return random.randint(0, 4)  # what is 4 supposed to represent?

roll_dice_literal = roll_dice

# Recommended
DICE_SIDES = 4

def roll_dice():
    return random.randint(0, DICE_SIDES)

random.seed(42)
print([roll_dice() for _ in range(5)])
```

## 5. Describe purpose, not implementation


```python
DICE_SIDES = 4

# Not recommended
def roll_dice_using_randint():
    return random.randint(0, DICE_SIDES)  
    
# Recommended
def roll_dice():
    return random.randint(0, DICE_SIDES)

random.seed(42)
print([roll_dice_using_randint() for _ in range(5)])
random.seed(42)
print([roll_dice() for _ in range(5)])
```

## 7. Separate Boolean-flag responsibilities


```python
text = "Python is a simple and elegant programming language."

# Not recommended
def transform_text(text, uppercase):
    if uppercase:
        return text.upper()
    else:
        return text.lower()

uppercase_text = transform_text(text, True)
lowercase_text = transform_text(text, False)


# Recommended
def transform_to_uppercase(text):
    return text.upper()

def transform_to_lowercase(text):
    return text.lower()

uppercase_text = transform_to_uppercase(text)
lowercase_text = transform_to_lowercase(text)

print(uppercase_text)
print(lowercase_text)
```

## 8. Avoid redundant class context


```python
# Not recommended
class Person:
    def __init__(self, person_username, person_email, person_phone, person_address):
        self.person_username = person_username
        self.person_email = person_email
        self.person_phone = person_phone
        self.person_address = person_address

PersonWithRedundantContext = Person

# Recommended
class Person:
    def __init__(self, username, email, phone, address):

        self.username = username
        self.email = email
        self.phone = phone
        self.address = address

person = Person("reader", "reader@example.test", "000", "Example Street")
print(person.username, person.email)
```

## 9. Aligned and hanging indentation


```python
var_one, var_two, var_three, var_four = 1, 2, 3, 4

# Supply the function referenced by the first alignment example.
def long_function_name(var_one, var_two, var_three, var_four):
    print(var_one)

# Correct:

# Aligned with opening delimiter.
foo = long_function_name(var_one, var_two,
                         var_three, var_four)

# Add 4 spaces (an extra level of indentation) to distinguish arguments from the rest.
def long_function_name(
        var_one, var_two, var_three,
        var_four):
    print(var_one)

# Hanging indents should add a level.
foo = long_function_name(
    var_one, var_two,
    var_three, var_four)
```

## 10. Intentionally nonrecommended indentation


```python
# Wrong:

# Arguments on first line forbidden when not using vertical alignment.
foo = long_function_name(var_one, var_two,
    var_three, var_four)

# Further indentation required as indentation is not distinguishable.
def long_function_name(
    var_one, var_two, var_three,
    var_four):
    print(var_one)

long_function_name(var_one, var_two, var_three, var_four)
```

## 11. Comments with local example context


```python
# Local in-memory stand-ins, not Django models or HTTP responses.
class LocalQuery:
    def __init__(self, found):
        self.found = found

    def exists(self):
        return self.found


class LocalUsers:
    def filter(self, *, id):
        return LocalQuery(id in {1})


class User:
    objects = LocalUsers()


class Response(dict):
    pass


def user_lookup(user_id):
    # This checks if the user with the given ID doesn't exist.
    if not User.objects.filter(id=user_id).exists():
        return Response({
            'detail': 'The user with this ID does not exist.',
        })
    return Response({"detail": "User exists."})

print(user_lookup(999))
print(user_lookup(1))
```

## 12. Single-line docstrings


```python
def population_density(population, land_area):
    """Calculate the population density of an area."""
    return population / land_area

population_density_single_line = population_density
print(population_density(1000, 25))
print(population_density.__doc__)
```

## 13. Multi-line docstrings


```python
def population_density(population, land_area):
    """Calculate the population density of an area.

    Args:
    population: int. The population of the area
    land_area: int or float. This function is unit-agnostic, if you pass in values in terms of square km or square miles the function will return a density in those units.

    Returns:
    population_density: population/land_area. The population density of a 
    particular area.
    """
    return population / land_area

population_density_multiline = population_density
print(population_density(1000, 25))
print(population_density.__doc__.splitlines()[0])
```

## Regression checks

Check retained values/signatures, deterministic random examples, before/after behavior, class fields, executable indentation, both local lookup branches and docstrings/zero-area errors.


```python
import inspect
import io
from contextlib import redirect_stdout

assert au == total_active_users == 105
assert c == cities_list == ["UK", "USA", "UAE"]
assert client_first_name == "John" and client_last_name == "Doe"
assert list(inspect.signature(fetch_clients).parameters) == ["response", "variable"]
assert list(inspect.signature(fetch_posts).parameters) == ["response", "variable"]
assert fetch_clients(None, None) is None
assert fetch_posts(None, None) is None
assert fetch_user(1) is None and get_post(1) is None and fetch_post(1) is None

# The source uses a zero-based inclusive randint range, not physical die faces.
for function in [roll_dice_literal, roll_dice_using_randint, roll_dice]:
    random.seed(42)
    assert [function() for _ in range(5)] == [0, 0, 2, 1, 1]
    assert all(0 <= function() <= DICE_SIDES for _ in range(100))

for value in ["", "Hello", "Straße", "MiXeD 123"]:
    assert transform_text(value, True) == transform_to_uppercase(value)
    assert transform_text(value, False) == transform_to_lowercase(value)

details = ("reader", "reader@example.test", "000", "Example Street")
old_person = PersonWithRedundantContext(*details)
new_person = Person(*details)
for field, expected in zip(["username", "email", "phone", "address"], details):
    assert getattr(old_person, "person_" + field) == getattr(new_person, field) == expected

# Both indentation demonstrations compile and run; one is intentionally poor style.
stream = io.StringIO()
with redirect_stdout(stream):
    assert long_function_name(1, 2, 3, 4) is None
assert stream.getvalue() == "1\n"

assert user_lookup(999) == {"detail": "The user with this ID does not exist."}
assert user_lookup(1) == {"detail": "User exists."}
for function in [population_density_single_line, population_density_multiline]:
    assert function(1000, 25) == 40.0
    assert function(0, 10) == 0.0
    assert function(3, 0.5) == 6.0
    assert inspect.getdoc(function).startswith("Calculate the population density of an area.")
    try:
        function(1, 0)
    except ZeroDivisionError:
        pass
    else:
        raise AssertionError("Zero area must not silently succeed")

print("All naming, formatting, text-transform, local-lookup and docstring checks passed.")
```
