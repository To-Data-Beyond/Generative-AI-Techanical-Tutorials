# Selecting & Replacing Values in pandas

Companion to the [complete article](https://todatabeyond.com/blog/selecting-and-replacing-values-in-pandas-dataframe-effectively). Run all cells in order with Python 3.11+ on CPU. No API keys or GPU. Setup installs pinned pandas 3.1.0, NumPy and certifi. Datasets download over verified HTTPS at an immutable original-author commit and are SHA-256 checked.

All 25 original examples remain. Assign directly to the DataFrame or assign the returned replacement Series/frame (no chained inplace writes). Every comparison starts from a fresh copy of the same original input; copying/downloading/displaying is outside measured intervals. Use high-resolution perf_counter and exact integer summation. Timing values/rankings are machine-dependent, not guaranteed speedups.

The original baby-name file contains 11,345 rows and no lowercase female category: the single-value replacement therefore correctly reports zero matches. A separate regression fixture exercises lowercase and missing-value handling without altering the real data. The original poker file has 25,010 hands, not all possible hands. The prose says three datasets but lists and uses two. Source prose/screenshots remain in the article; current Output panels come from actual execution.

Warnings for chained assignments and future API problems are errors during execution. Regression tests check full-data results and that loc label semantics differ from iloc positional semantics on a relabeled index. Outputs and counts are cleared before committing.

```python
import hashlib
import io
import ssl
import warnings
from urllib.request import urlopen
import certifi
import pandas as pd

DATASET_BASE = "https://raw.githubusercontent.com/youssefHosni/Efficient-Python-for-Data-Scientists-Book/2507c7f72e3cb08ef349047a4ca052bcc7307337/Datasets/"
DATASET_HASHES = {
    "poker_hand.csv": "0778285d95b3360696e9e3d903b2d9d230832b771586da9d8a3459384897db20",
    "Popular_Baby_Names.csv": "5e41e47c4b8a01901ef9a5eab2e17df35c8b4e29fdee7a063a3045094f1109bb",
}

def load_dataset(filename):
    context = ssl.create_default_context(cafile=certifi.where())
    with urlopen(DATASET_BASE + filename, context=context, timeout=60) as response:
        payload = response.read()
    assert hashlib.sha256(payload).hexdigest() == DATASET_HASHES[filename]
    frame = pd.read_csv(io.BytesIO(payload))
    print(filename, "shape:", frame.shape)
    return frame

warnings.filterwarnings("error", category=pd.errors.ChainedAssignmentError)
warnings.filterwarnings("error", category=FutureWarning)
print("Dataset loader ready; HTTPS and SHA-256 verified, chained-assignment warnings treated as errors.")
```

## Example 7


```python
def sum_brute_force(N):
  res = 0
  for i in range(1,N+1):
    res+=i
  return res

# Using brute force
bf_start_time = time.perf_counter()
bf_result = sum_brute_force(1000000)
bf_end_time = time.perf_counter()

print("Time using brute force: {} sec".format(bf_end_time - bf_start_time))
print("Sum:", bf_result)
```

## Example 8


```python
def sum_formula(N):
  return N*(N+1)//2
  
# Using the formula
formula_start_time = time.perf_counter()
formula_result = sum_formula(1000000)
formula_end_time = time.perf_counter()

print("Time using the formula: {} sec".format(formula_end_time - formula_start_time))
print("Sum:", formula_result)
```

```python
names_loc = names_source.copy(deep=True)
gender_mask = names_loc['Gender'].eq('female')
start_time = time.perf_counter()
names_loc.loc[gender_mask, 'Gender'] = 'FEMALE'
end_time = time.perf_counter()
pandas_time = end_time - start_time
print("Replace values using .loc[]: {} sec".format(pandas_time))
print("Matched rows:", int(gender_mask.sum()))
print(names_loc['Gender'].value_counts().to_string())
```

## Regression checks


```python
from pandas.testing import assert_frame_equal

assert version('pandas') == '3.1.0'
assert poker_data.shape == (25010, 11)
assert names_source.shape == (11345, 6)
assert names_source['Gender'].value_counts().to_dict() == {'FEMALE': 5806, 'MALE': 5539}
assert names_source['Ethnicity'].value_counts().to_dict() == {
    'WHITE NON HISPANIC': 3365, 'HISPANIC': 3314,
    'BLACK NON HISPANIC': 1704, 'ASIAN AND PACIFIC ISLANDER': 1662,
    'WHITE NON HISP': 637, 'ASIAN AND PACI': 335, 'BLACK NON HISP': 328}
assert squares_comprehension == squares_loop
assert len(squares_loop) == 1000000
assert squares_loop[-1] == 999998000001
assert bf_result == formula_result == 500000500000
assert isinstance(sum_formula(10**20), int)
assert sum_formula(10**20) == (10**20)*(10**20+1)//2
for n in [0, 1, 2, 99]:assert sum_brute_force(n) == sum_formula(n)
assert_frame_equal(selected_loc, selected_iloc)
assert_frame_equal(selected_loc, poker_data.head(500))
assert_frame_equal(columns_iloc, columns_named)
assert columns_named.columns.tolist() == ['S1', 'R1', 'S2']
assert columns_named.shape == (25010, 3)
assert int(gender_mask.sum()) == 0
assert_frame_equal(names_loc, names_replace)
assert_frame_equal(names_loc, names_source)
assert_frame_equal(ethnicity_loc, ethnicity_replace)
assert int(white_mask.sum()) == 4002
assert ethnicity_replace['Ethnicity'].eq('WNH').sum() == 4002
assert not ethnicity_replace['Ethnicity'].isin(['WHITE NON HISPANIC', 'WHITE NON HISP']).any()
assert_frame_equal(names_dict, names_multiple)
assert names_dict['Gender'].value_counts().to_dict() == {'GIRL': 5806, 'BOY': 5539}
assert names_nested['Ethnicity'].value_counts().to_dict() == {
    'WHITE': 4002, 'HISPANIC': 3314, 'BLACK': 2032, 'ASIAN': 1997}
assert names_nested['Ethnicity'].eq('HISPANIC').sum() == names_source['Ethnicity'].eq('HISPANIC').sum()
assert_frame_equal(names_nested.drop(columns='Ethnicity'), names_source.drop(columns='Ethnicity'))
assert_frame_equal(names, names_source)

# The real CSV has no lowercase female values. Exercise that branch separately.
fixture = pd.DataFrame({'Gender': ['female', 'FEMALE', 'MALE', None],
                        'Ethnicity': ['WHITE NON HISP', 'HISPANIC', 'BLACK NON HISP', None]})
left = fixture.copy(deep=True)
left.loc[left['Gender'].eq('female'), 'Gender'] = 'FEMALE'
right = fixture.copy(deep=True)
right['Gender'] = right['Gender'].replace('female', 'FEMALE')
assert_frame_equal(left, right)
assert left['Gender'].iloc[:3].tolist() == ['FEMALE', 'FEMALE', 'MALE']
assert pd.isna(left['Gender'].iloc[3])
assert fixture['Gender'].iloc[0] == 'female'

# loc uses labels, iloc uses positions; equivalence is not universal.
relabeled = poker_data.copy(deep=True)
relabeled.index = relabeled.index + 100000
assert relabeled.iloc[:500].shape == (500, 11)
try:
    relabeled.loc[range(500)]
except KeyError:
    pass
else:
    raise AssertionError('loc must look up labels, not row positions')
assert_frame_equal(relabeled.loc[range(100000, 100500)], relabeled.iloc[:500])
print('All full-dataset selection, replacement, exact-sum, input-preservation and copy-on-write checks passed.')
print('Poker rows:', len(poker_data), 'baby-name rows:', len(names_source))
print('White ethnicity replacements:', int(white_mask.sum()))
print('Dictionary gender counts:', names_dict['Gender'].value_counts().to_dict())
print('Nested ethnicity counts:', names_nested['Ethnicity'].value_counts().to_dict())
```
