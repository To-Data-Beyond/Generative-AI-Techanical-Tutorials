## Example 6

````python
import hashlib
import io
import ssl
import urllib.request
DATA_URL="https://raw.githubusercontent.com/youssefHosni/Efficient-Python-for-Data-Scientists/2507c7f72e3cb08ef349047a4ca052bcc7307337/Datasets/Popular_Baby_Names.csv"
DATA_SHA256="5e41e47c4b8a01901ef9a5eab2e17df35c8b4e29fdee7a063a3045094f1109bb"
with urllib.request.urlopen(DATA_URL,context=ssl.create_default_context(cafile=certifi.where()),timeout=60) as response:data=response.read()
assert hashlib.sha256(data).hexdigest()==DATA_SHA256
Popular_baby_names_df=pd.read_csv(io.BytesIO(data))
print("Full dataset shape:",Popular_baby_names_df.shape)
print(Popular_baby_names_df.head().to_string(index=False))
````

## Example 7

````python
from pathlib import Path
profile=pp.ProfileReport(Popular_baby_names_df,title="Pandas Profiling Report",pool_size=1,progress_bar=False)
profile.to_file("popular_baby_names_report.html")
description=profile.get_description()
print("Report rows:",description.table["n"],"| columns:",description.table["n_var"])
print("Missing cells:",description.table["n_cells_missing"],"| duplicate rows:",description.table["n_duplicates"])
print("Report variables:",list(description.variables))
print("Correlation matrices:",list(description.correlations))
print("HTML written:",Path("popular_baby_names_report.html").exists())
````

## Independent regression checks

````python
import json
df=Popular_baby_names_df
assert df.shape[1]==6 and len(df)>10000
assert description.table["n"]==len(df) and description.table["n_var"]==len(df.columns)
assert description.table["n_cells_missing"]==int(df.isna().sum().sum())
assert description.table["n_duplicates"]==int(df.duplicated().sum())
for name in ["Year of Birth","Count","Rank"]:
    v=description.variables[name]
    np.testing.assert_allclose(v["mean"],df[name].mean(),rtol=1e-12)
    assert v["min"]==df[name].min() and v["max"]==df[name].max()
    assert v["n_missing"]==int(df[name].isna().sum())
assert set(description.variables)==set(df.columns)
html=Path("popular_baby_names_report.html").read_text()
for section in ["Overview","Variables","Interactions","Correlations","Missing values","Sample"]:assert section.lower() in html.lower(),section
profile.to_file("popular_baby_names_report.json")
payload=json.loads(Path("popular_baby_names_report.json").read_text())
assert payload["table"]["n"]==len(df)
print("All original rows profiled; independent pandas counts/statistics and HTML/JSON exports verified.")
print("Original data SHA-256:",DATA_SHA256)
print("Count/Rank Pearson correlation:",f"{df['Count'].corr(df['Rank']):.6f}")
````
