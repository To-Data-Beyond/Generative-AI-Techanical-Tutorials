## Before you begin

This is a documentation-first local companion. Claude Code and the Z.ai Coding Tool Helper run in your terminal, not inside a hosted notebook runtime.

- Use Node.js 18 or newer.
- Work only inside a project directory you trust.
- Obtain the API key from your own Z.ai account.
- Never paste a real key into this notebook or commit one to Git.
- This notebook intentionally contains no credentials or saved execution outputs.


> **Archive update — August 2026:** The article was originally published for GLM-5 in February 2026. Z.ai changes supported models, default mappings, quotas, and prices over time. Verify the current official documentation before using these settings. The original 60% headline is not a universal savings guarantee.


## 1. Install or update Claude Code

Run these commands in a local terminal:

```bash
npm install -g @anthropic-ai/claude-code
claude --version
claude update
```

Start Claude Code from the project you want it to inspect, and review the workspace-trust prompt before granting access.


## 2. Configure the GLM Coding Plan

The official Z.ai helper guides you through selecting a regional plan, entering your API key locally, and loading the plan into Claude Code:

```bash
npx @z_ai/coding-helper
```

After synchronization, fully close Claude Code and open a new terminal so the new environment is loaded.


## 3. Verify the active route

Inside a fresh Claude Code session, inspect the status and model menu:

```text
/status
/model
```

Z.ai notes that Claude Code can display a Claude alias while the configured mapping routes it to a GLM model. When troubleshooting, exit Claude Code and run:

```bash
npx @z_ai/coding-helper doctor
```


## 4. Manual configuration template

If the helper cannot configure Claude Code, Z.ai documents this structure for `~/.claude/settings.json`. Keep the placeholder in this notebook and replace it only in your private local file.


````python
import json

settings_template = {
    "env": {
        "ANTHROPIC_AUTH_TOKEN": "your_zai_api_key",
        "ANTHROPIC_BASE_URL": "https://api.z.ai/api/anthropic",
        "API_TIMEOUT_MS": "3000000",
    }
}

assert settings_template["env"]["ANTHROPIC_AUTH_TOKEN"] == "your_zai_api_key"
assert settings_template["env"]["ANTHROPIC_BASE_URL"].startswith("https://")
print(json.dumps(settings_template, indent=2))
````

Copy the validated structure into your private `~/.claude/settings.json`, replace the placeholder locally, save valid JSON, and restart Claude Code. Do not save or commit the real key in this notebook.

Avoid hardcoding model mappings unless you need a specific supported model. Removing stale mapping variables lets Z.ai's plan defaults follow newer supported models.


## 5. Return to Anthropic's service

Remove the Z.ai-specific values from the `env` object in `~/.claude/settings.json`, then fully restart Claude Code:

```text
ANTHROPIC_AUTH_TOKEN
ANTHROPIC_BASE_URL
API_TIMEOUT_MS
ANTHROPIC_DEFAULT_OPUS_MODEL
ANTHROPIC_DEFAULT_SONNET_MODEL
ANTHROPIC_DEFAULT_HAIKU_MODEL
```

Review the complete `env` object. Leaving the custom base URL or a stale mapping behind can cause confusing authentication or model-selection behavior.


## 6. Suggested routing policy

Use a lower-cost GLM route for measurable, lower-risk tasks such as planning, draft code, routine documentation, and iterations covered by tests. Escalate ambiguous architecture decisions, large migrations, production-critical work, and expert audits to a premium frontier model when the additional capability materially reduces risk.

Compare models using successful completion rate, review effort, latency, and total token usage—not token price alone.
