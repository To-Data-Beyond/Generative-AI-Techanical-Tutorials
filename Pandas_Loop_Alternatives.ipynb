```python
import hashlib
import io
import ssl
import warnings
from urllib.request import urlopen
import certifi
import numpy as np
import pandas as pd

POKER_URL = "https://raw.githubusercontent.com/youssefHosni/Efficient-Python-for-Data-Scientists-Book/2507c7f72e3cb08ef349047a4ca052bcc7307337/Datasets/poker_hand.csv"
POKER_SHA256 = "0778285d95b3360696e9e3d903b2d9d230832b771586da9d8a3459384897db20"

def load_poker_data():
    context = ssl.create_default_context(cafile=certifi.where())
    with urlopen(POKER_URL, context=context, timeout=60) as response:
        payload = response.read()
    assert hashlib.sha256(payload).hexdigest() == POKER_SHA256
    frame = pd.read_csv(io.BytesIO(payload))
    print("poker_hand.csv shape:", frame.shape)
    return frame

warnings.filterwarnings("error", category=FutureWarning)
print("Poker loader ready; HTTPS and SHA-256 are verified.")
```

```python
poker_data = load_poker_data()
poker_original = poker_data.copy(deep=True)
rank_columns = ["R1", "R2", "R3", "R4", "R5"]
print(poker_data.head().to_string(index=False))
```

```python
# Using For loop

for_loop_start_time= time.perf_counter()
result=[]
for i in range(0,1000000):
  result.append(i*i)
for_loop_end_time= time.perf_counter()
print("Time using the for loop: {} sec".format(for_loop_end_time - for_loop_start_time))
squares_loop = result
print("Squares:", len(result), result[:3], result[-3:])
```

## Example 6


```python
def sum_brute_force(N):
  res = 0
  for i in range(1,N+1):
    res+=i
  return res

# Using brute force
bf_start_time = time.perf_counter()
bf_result = sum_brute_force(1000000)
bf_end_time = time.perf_counter()

print("Time using brute force: {} sec".format(bf_end_time - bf_start_time))
print("Sum:", bf_result)
```

## Example 7


```python
def sum_formula(N):
  return N*(N+1)//2
  
# Using the formula
formula_start_time = time.perf_counter()
formula_result = sum_formula(1000000)
formula_end_time = time.perf_counter()

print("Time using the formula: {} sec".format(formula_end_time - formula_start_time))
print("Sum:", formula_result)
```

## Regression checks


```python
from types import GeneratorType
from pandas.testing import assert_frame_equal, assert_series_equal

assert version('pandas') == '3.1.0'
assert poker_data.shape == (25010, 11)
assert rank_columns == ['R1', 'R2', 'R3', 'R4', 'R5']
assert squares_comprehension == squares_loop
assert len(squares_loop) == 1000000 and squares_loop[-1] == 999998000001
assert bf_result == formula_result == 500000500000
for n in [0, 1, 2, 99]:assert sum_brute_force(n) == sum_formula(n)
assert isinstance(sum_formula(10**20), int)
assert sum_formula(10**20) == (10**20)*(10**20+1)//2
assert isinstance(city_name_generator(), GeneratorType)
assert list(city_name_generator()) == ['New York', 'London', 'Tokyo', 'Sao Paolo']
try:
    next(city_names)
except StopIteration:
    pass
else:
    raise AssertionError('All four original generator values must have been consumed')
assert index == len(poker_data) - 1
np.testing.assert_allclose(data_sqrt.to_numpy(), np.sqrt(poker_data.to_numpy()))
assert data_sqrt.shape == poker_data.shape
expected_rows = poker_data[rank_columns].to_numpy().sum(axis=1)
for values in [row_sums_apply, row_sums_loop, row_sums_vectorized,
               row_sums_iterrows, row_sums_apply_again, row_sums_numpy,
               row_sums_pandas]:
    np.testing.assert_array_equal(values, expected_rows)
    assert len(values) == 25010
assert expected_rows[:5].tolist() == [47, 47, 47, 47, 47]
assert (int(expected_rows.min()), int(expected_rows.max())) == (8, 62)
expected_columns = {'R1': 174951, 'R2': 175425, 'R3': 175424, 'R4': 173631, 'R5': 174138}
assert column_sums_apply.to_dict() == column_sums_pandas.to_dict() == expected_columns
assert_series_equal(column_sums_apply, column_sums_pandas)
# Catch the source's omitted fifth rank and duplicate-column comparison defects.
wrong_four_card_sums = poker_data[['R1', 'R2', 'R3', 'R4']].sum(axis=1)
np.testing.assert_array_equal(expected_rows - wrong_four_card_sums, poker_data['R5'])
assert (expected_rows > wrong_four_card_sums).all()
assert poker_data[['R1', 'R1', 'R3', 'R4', 'R5']].columns.tolist() != rank_columns
assert_frame_equal(poker_data, poker_original)
assert_frame_equal(data_sqrt.pow(2).round().astype('int64'), poker_original)
assert all(duration >= 0 for duration in [for_loop_time, apply_time, pandas_time, vectorization_time, iterrows_time])
assert_frame_equal(poker_data, poker_original)
print('All full-dataset row/column sum, square-root, generator, exact-integer and input-preservation checks passed.')
print('Poker records:', len(poker_data))
print('First five hand rank sums:', expected_rows[:5].tolist())
print('Rank column sums:', expected_columns)
print('All seven row-sum results agree across 25,010 hands.')
```
