`````python
from pathlib import Path
import tempfile,shutil,subprocess,shlex,hashlib
workspace = Path(tempfile.mkdtemp(prefix='tdb-claude-usage-'))
print('Created an owned temporary project; no model session started.')
`````

## Store the project instructions

This is the article’s fictional FastAPI/PostgreSQL/Redis project, not an application supplied with this notebook. Its pytest/ruff commands describe that fictional project and are not run against an absent application.

`````python
project_instructions = '# Project Instructions\n\n## Project overview\nThis is a FastAPI + PostgreSQL + Redis backend for an agentic RAG application.\n\n## Commands\n- Run tests with: `pytest`\n- Run linting with: `ruff check .`\n- Run formatting with: `ruff format .`\n\n## Coding rules\n- Use type hints in all Python functions.\n- Keep API routes thin.\n- Put business logic inside services.\n- Do not create new abstractions unless they reduce real duplication.\n\n## AI workflow rules\n- Before making large changes, propose a plan first.\n- After editing code, explain what changed and what still needs testing.\n- When I correct a repeated mistake, save it to memory.'
instructions_path = workspace / 'CLAUDE.md'
instructions_path.write_text(project_instructions + '\n', encoding='utf-8')
assert instructions_path.read_text(encoding='utf-8').rstrip() == project_instructions
sections = [line for line in project_instructions.splitlines() if line.startswith('## ')]
assert sections == ['## Project overview', '## Commands', '## Coding rules', '## AI workflow rules']
assert len(project_instructions.splitlines()) < 200
assert all(command in project_instructions for command in ['pytest','ruff check .','ruff format .'])
print('Project overview, commands, coding rules and AI workflow rules retained.')
`````

`````python
restart_prompt = 'I want to move this task to a new chat.\n\nPlease summarize only the important context needed to continue:\n- The goal of the task\n- The decisions we already made\n- The current state\n- The constraints and preferences\n- The next steps\nWrite it as a prompt I can paste into a new Claude chat.'
expected = ['The goal of the task', 'The decisions we already made', 'The current state', 'The constraints and preferences', 'The next steps']
assert all(item in restart_prompt for item in expected)
(workspace / 'restart-prompt.txt').write_text(restart_prompt + '\n', encoding='utf-8')
digest = hashlib.sha256(instructions_path.read_bytes()).hexdigest()
assert instructions_path.parent == workspace
print('Restart prompt includes all five handoff fields; project instructions unchanged.')
`````

`````python
# Inspect documented flags only. Never run the launch command in this notebook.
launch = ['claude', '--permission-mode', 'plan']
assert shlex.split(shlex.join(launch)) == launch
executable = shutil.which('claude')
if executable:
    version = subprocess.run([executable,'--version'],check=True,capture_output=True,text=True,timeout=30)
    help_result = subprocess.run([executable,'--help'],check=True,capture_output=True,text=True,timeout=30)
    assert '--permission-mode' in help_result.stdout and 'plan' in help_result.stdout
    print('Installed Claude Code:', version.stdout.strip())
    print('Plan flag is advertised by local CLI help; no account or permission behavior tested.')
else:
    print('Claude CLI not installed; optional help inspection skipped.')
assert hashlib.sha256(instructions_path.read_bytes()).hexdigest() == digest
print('No model calls, external messages, schedules or real Claude configuration changes.')
`````

`````python
# Remove only these two fixture files and the exact owned empty directory.
for name in ['CLAUDE.md', 'restart-prompt.txt']:
    (workspace / name).unlink()
workspace.rmdir()
assert not workspace.exists()
print('Owned fixture cleaned up.')
`````
