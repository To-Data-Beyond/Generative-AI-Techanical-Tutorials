# 11 Docker Container Images for Generative AI & ML Projects

Companion notes for the To Data & Beyond archive article. These cells document the image choices and commands; run them from a host with Docker installed. Pin image tags for reproducible production workloads.

## Machine learning and data science

```python
# Python + NumPy + pandas
docker build -t ml-python - <<'EOF'
FROM python:3.8
RUN pip install --no-cache-dir numpy pandas
EOF
```

```python
# Jupyter data-science stack
docker run -it --rm -p 8888:8888 jupyter/datascience-notebook
```

## Generative AI and deep learning

```python
# Hugging Face Transformers
docker build -t hf-transformers - <<'EOF'
FROM huggingface/transformers-pytorch-gpu
RUN python main.py
EOF
```

```python
# CUDA runtime base
docker build -t cuda-runtime - <<'EOF'
FROM nvidia/cuda:12.2.0-runtime-ubuntu22.04
RUN python main.py
EOF
```

```python
# TensorFlow
docker build -t tensorflow-app - <<'EOF'
FROM tensorflow/tensorflow:latest
RUN python main.py
EOF
```

```python
# PyTorch
docker build -t pytorch-app - <<'EOF'
FROM pytorch/pytorch:latest
RUN python main.py
EOF
```

```python
# Ollama
docker run -it --rm ollama/ollama
```

```python
# Qdrant vector database
docker run -it --rm -p 6333:6333 qdrant/qdrant
```

## Workflow orchestration and ML lifecycle

```python
# Airflow
docker run -it --rm -p 8080:8080 apache/airflow
```

```python
# MLflow
docker run -it --rm -p 9000:9000 ghcr.io/mlflow/mlflow
```

```python
# Kubeflow Notebooks
docker run -it --rm -p 8888:8888 kubeflownotebookswg/jupyter-pytorch
```

## Reproducibility checklist

- Prefer versioned tags instead of `latest`.
- Match CUDA, driver, and framework versions before enabling GPUs.
- Mount persistent volumes for notebooks, model caches, and experiment artifacts.
- Review image provenance and scan images before production deployment.
