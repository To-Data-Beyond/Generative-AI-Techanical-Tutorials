## 1. Reverse a list in place


```python
my_list = [1, 2, 3, 4, 5]

# Reversing a list in place
my_list.reverse()
print(my_list)
```

## 2. Enumerate indices and values


```python
# Sample list
my_list = ['apple', 'banana', 'orange', 'grape']

# Looping through the elements and their indices using enumerate
for index, value in enumerate(my_list):
    print(f"Index: {index}, Value: {value}")
```

## 3. Unpack a tuple


```python
# Sample tuple
my_tuple = (1, 2, 3)

# Unpacking the tuple into individual variables
a, b, c = my_tuple

# Displaying the values
print(f"a: {a}, b: {b}, c: {c}")
```

## 4. Unpack returned values


```python
# Sample function returning multiple values
def get_coordinates():
    return 4, 5, 6

# Unpacking the returned values
x, y, z = get_coordinates()

print((x, y, z))
```

## 5. Swap two variables


```python
a = 10
b = 20

# Swapping values
a, b = b, a

print((a, b))
```

## 6. Extended unpacking


```python
# Sample list
my_list = [1, 2, 3, 4, 5]

# Unpacking the first element and collecting the rest
first, *rest = my_list

print(first, rest)
```

## 7. List comprehension


```python
# Traditional loop
squares = []
for x in range(5):
    squares.append(x**2)

# List comprehension equivalent
squares = [x**2 for x in range(5)]

print(squares)
```

## 8. Conditional list comprehension


```python
# Traditional loop with condition
even_squares = []
for x in range(5):
    if x % 2 == 0:
        even_squares.append(x**2)

# List comprehension equivalent
even_squares = [x**2 for x in range(5) if x % 2 == 0]

print(even_squares)
```

## 9. Dictionary comprehension


```python
# Traditional loop
squares_dict = {}
for x in range(5):
    squares_dict[x] = x**2

# Dictionary comprehension equivalent
squares_dict = {x: x**2 for x in range(5)}

print(squares_dict)
```

## 10. Conditional dictionary comprehension


```python
# Traditional loop with condition
even_squares_dict = {}
for x in range(5):
    if x % 2 == 0:
        even_squares_dict[x] = x**2

# Dictionary comprehension equivalent
even_squares_dict = {x: x**2 for x in range(5) if x % 2 == 0}

print(even_squares_dict)
```

## 11. Build a dictionary with zip


```python
keys = ['a', 'b', 'c']
values = [1, 2, 3]

# Dictionary comprehension using zip
combined_dict = {key: value for key, value in zip(keys, values)}

print(combined_dict)
```

## 12. Filter dictionary keys


```python
# Include only even values in the dictionary
my_dict = {x: x**2 for x in range(10) if x % 2 == 0}

print(my_dict)
```

## 13. Return a sorted copy


```python
my_list = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]

sorted_list = sorted(my_list)
print(sorted_list)
```

## 14. Sort a list in place


```python
my_list = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]

my_list.sort()
print(my_list)
```

## 15. Descending sort


```python
my_list = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]

descending_sorted = sorted(my_list, reverse=True)
print(descending_sorted)
```

## 16. Sort by a custom key


```python
words = ['apple', 'banana', 'orange', 'grape']

# Sort by length
sorted_by_length = sorted(words, key=len)
print(sorted_by_length)
```

## 17. Sort objects by an attribute


```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

people = [Person('Alice', 25), Person('Bob', 30), Person('Charlie', 22)]

# Sort people by age
sorted_people = sorted(people, key=lambda person: person.age)

print([(person.name, person.age) for person in sorted_people])
```

## 18. Sort with multiple criteria


```python
records = [('Alice', 25), ('Bob', 30), ('Alice', 22)]

# Sort by name, then by age
sorted_records = sorted(records, key=lambda x: (x[0], x[1]))

print(sorted_records)
```
