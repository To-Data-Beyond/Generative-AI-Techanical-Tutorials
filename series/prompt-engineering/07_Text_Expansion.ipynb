`````python
import os, json, re, sys
from importlib.metadata import version
from openai import OpenAI
from dotenv import load_dotenv
load_dotenv()
MODEL = os.getenv("MODEL", "gpt-5.4-mini")
client = OpenAI(max_retries=0, timeout=60)
def get_completion_from_messages(messages, model=MODEL, temperature=None):
    options = {} if temperature is None else {"temperature": temperature}
    result = client.chat.completions.create(
        model=model, messages=messages, reasoning_effort="none",
        max_completion_tokens=2048, **options)
    if result.choices[0].finish_reason != "stop":
        raise RuntimeError("The completion did not finish normally")
    content = result.choices[0].message.content
    if not content:
        raise RuntimeError("No text returned")
    return content
def get_completion(prompt, model=MODEL, temperature=None):
    return get_completion_from_messages(
        [{"role": "user", "content": prompt}], model=model, temperature=temperature)
print("Python:", sys.version.split()[0])
print("openai:", version("openai"), "python-dotenv:", version("python-dotenv"))
print("Model:", MODEL)
`````

`````python
prompt = f"""
You are a customer service AI assistant.
Your task is to draft an email reply to a valued customer.
Given the customer email delimited by ```,  Generate a reply to thank the customer for their review.
If the sentiment is positive or neutral, thank them for  their review.
If the sentiment is negative, apologize and suggest that  they can reach out to customer service. 
Make sure to use specific details from the review.
Write in a concise and professional tone.
Sign the email as `AI customer agent`.
Customer review: ```{review}```
Review sentiment: {sentiment}
"""
response = get_completion(prompt, temperature=0)
print(response)
email_low = response
`````

`````python
prompt = f"""
You are a customer service AI assistant.
Your task is to draft an email reply to a valued customer.
Given the customer email delimited by ```,  Generate a reply to thank the customer for their review.
If the sentiment is positive or neutral, thank them for  their review.
If the sentiment is negative, apologize and suggest that  they can reach out to customer service. 
Make sure to use specific details from the review.
Write in a concise and professional tone.
Sign the email as `AI customer agent`.
Customer review: ```{review}```
Review sentiment: {sentiment}
"""
response = get_completion(prompt, temperature=0.7)
print(response)
email_varied = response
`````

`````python
assert sentiment == "negative" and len(review.split()) > 250
for email in [email_low, email_varied]:
    lower = email.lower()
    assert "ai customer agent" in lower
    assert "thank" in lower and ("sorry" in lower or "apolog" in lower)
    assert "customer service" in lower
    assert any(term in lower for term in ["motor", "warranty", "pricing", "price"])
    assert not re.search(r"https?://|[A-Z0-9._%+-]+@[A-Z0-9.-]+\.[A-Z]{2,}", email, re.I)
print("PASS: full source review, AI attribution, negative-sentiment apology, source details and customer-service direction in both drafts. No email is sent.")
print("Drafts differ:", email_low != email_varied)
`````
