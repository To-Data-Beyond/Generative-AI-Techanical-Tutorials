````python
class HashTable:
    def __init__(self, size):
        if not isinstance(size, int) or size <= 0:
            raise ValueError("size must be a positive integer")
        self.size = size
        self.table = [None] * size

    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        if self.table[index] is None:
            self.table[index] = [(key, value)]
        else:
            for i, (existing_key, _) in enumerate(self.table[index]):
                if existing_key == key:
                    self.table[index][i] = (key, value)
                    break
            else:
                self.table[index].append((key, value))

    def get(self, key):
        index = self.hash_function(key)
        if self.table[index] is not None:
            for existing_key, value in self.table[index]:
                if existing_key == key:
                    return value
        return None

table = HashTable(3)
table.insert(1, "one"); table.insert(4, "four"); table.insert(1, "updated")
print("Hash collision / update:", table.get(4), table.get(1))
````

````python
class HashTableLinearProbing:
    def __init__(self, size):
        if not isinstance(size, int) or size <= 0:
            raise ValueError("size must be a positive integer")
        self.size = size
        self.table = [None] * size

    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        for _ in range(self.size):
            if self.table[index] is None or self.table[index][0] == key:
                self.table[index] = (key, value)
                return
            index = (index + 1) % self.size
        raise OverflowError("hash table is full")

    def get(self, key):
        index = self.hash_function(key)
        initial_index = index
        while self.table[index] is not None:
            existing_key, value = self.table[index]
            if existing_key == key:
                return value
            index = (index + 1) % self.size
            if index == initial_index:
                break
        return None

probing = HashTableLinearProbing(3)
probing.insert(1, "one"); probing.insert(4, "four"); probing.insert(1, "updated")
print("Linear probing:", probing.get(1), probing.get(4))
````

````python
class MaxHeap:
    def __init__(self):
        self.heap = []

    def insert(self, value):
        self.heap.append(value)
        self._heapify_up(len(self.heap) - 1)

    def extract_max(self):
        if len(self.heap) == 0:
            return None

        if len(self.heap) == 1:
            return self.heap.pop()

        max_value = self.heap[0]
        self.heap[0] = self.heap.pop()
        self._heapify_down(0)
        return max_value

    def _heapify_up(self, index):
        parent_index = (index - 1) // 2
        while index > 0 and self.heap[index] > self.heap[parent_index]:
            self.heap[index], self.heap[parent_index] = self.heap[parent_index], self.heap[index]
            index = parent_index
            parent_index = (index - 1) // 2

    def _heapify_down(self, index):
        left_child_index = 2 * index + 1
        right_child_index = 2 * index + 2
        largest = index

        if left_child_index < len(self.heap) and self.heap[left_child_index] > self.heap[largest]:
            largest = left_child_index

        if right_child_index < len(self.heap) and self.heap[right_child_index] > self.heap[largest]:
            largest = right_child_index

        if largest != index:
            self.heap[index], self.heap[largest] = self.heap[largest], self.heap[index]
            self._heapify_down(largest)

maximum_heap = MaxHeap()
for value in [3, 1, 4, 1, 5]: maximum_heap.insert(value)
print("Max heap extraction:", [maximum_heap.extract_max() for _ in range(5)])
````

````python
import heapq

class MinHeap:
    def __init__(self):
        self.heap = []

    def insert(self, value):
        heapq.heappush(self.heap, value)

    def extract_min(self):
        if len(self.heap) == 0:
            return None
        return heapq.heappop(self.heap)

minimum_heap = MinHeap()
for value in [3, 1, 4, 1, 5]: minimum_heap.insert(value)
print("Min heap extraction:", [minimum_heap.extract_min() for _ in range(5)])
````

````python
class ListNode:
    def __init__(self, value=0, next=None):
        self.value = value
        self.val = value
        self.next = next

def make_list(values):
    head = None
    for value in reversed(values):
        head = ListNode(value, head)
    return head

def list_values(head):
    result = []
    while head:
        result.append(head.value)
        head = head.next
    return result
print("Linked-list node and construction helpers ready.")

import heapq

def merge_k_sorted_lists(lists):
    min_heap = []
    for i, lst in enumerate(lists):
        if lst:
            heapq.heappush(min_heap, (lst.val, i, lst))

    dummy = ListNode()
    current = dummy

    while min_heap:
        val, i, node = heapq.heappop(min_heap)
        current.next = node
        current = current.next
        if node.next:
            heapq.heappush(min_heap, (node.next.val, i, node.next))

    return dummy.next

heads = [make_list([1, 4, 5]), make_list([1, 3, 4]), make_list([2, 6])]
print("Merged linked lists:", list_values(merge_k_sorted_lists(heads)))
````

````python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_end_of_word = False

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        node = self.root
        for char in word:
            if char not in node.children:
                node.children[char] = TrieNode()
            node = node.children[char]
        node.is_end_of_word = True

    def search(self, word):
        node = self.root
        for char in word:
            if char not in node.children:
                return False
            node = node.children[char]
        return node.is_end_of_word

    def starts_with_prefix(self, prefix):
        node = self.root
        for char in prefix:
            if char not in node.children:
                return False
            node = node.children[char]
        return True

trie = Trie()
for word in ["car", "card", "care", "cat"]: trie.insert(word)
print("Search / prefix:", trie.search("car"), trie.search("ca"), trie.starts_with_prefix("ca"))
````

````python
def autocomplete(trie, prefix):
    node = trie.root
    for char in prefix:
        if char not in node.children:
            return []
        node = node.children[char]
    suggestions = []
    def visit(current, suffix):
        if current.is_end_of_word:
            suggestions.append(prefix + "".join(suffix))
        for char, child in current.children.items():
            suffix.append(char)
            visit(child, suffix)
            suffix.pop()
    visit(node, [])
    return suggestions

print("Autocomplete:", autocomplete(trie, "car"))
````

````python
def longest_word(words):
    trie = Trie()
    for word in words:
        trie.insert(word)
    longest = ""
    path = []
    def visit(node):
        nonlocal longest
        if node.is_end_of_word:
            word = "".join(path)
            if len(word) > len(longest) or (len(word) == len(longest) and word < longest):
                longest = word
        for char, child in node.children.items():
            path.append(char)
            visit(child)
            path.pop()
    visit(trie.root)
    return longest

print("Longest dictionary word:", longest_word(["car", "card", "care", "cat"]))
````

````python
def frequency_counter(arr):
    frequency_map = {}
    for element in arr:
        if element in frequency_map:
            frequency_map[element] += 1
        else:
            frequency_map[element] = 1
    return frequency_map

print("Frequencies:", frequency_counter(["a", "b", "a", "c", "b", "a"]))
````

````python
def first_non_repeating_char(s):
    char_count = {}
    for char in s:
        char_count[char] = char_count.get(char, 0) + 1

    for char in s:
        if char_count[char] == 1:
            return char
    return None

print("First non-repeating character:", first_non_repeating_char("swiss"))
````

````python
def fibonacci(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("n must be a nonnegative integer")
    memo = {0: 0, 1: 1}
    def solve(k):
        if k not in memo:
            memo[k] = solve(k - 1) + solve(k - 2)
        return memo[k]
    return solve(n)

print("Fibonacci(30):", fibonacci(30))
````

````python
def length_of_lis(nums):
    if not nums:
        return 0

    dp = [1] * len(nums)

    for i in range(1, len(nums)):
        for j in range(i):
            if nums[i] > nums[j]:
                dp[i] = max(dp[i], dp[j] + 1)

    return max(dp)

print("Longest increasing subsequence length:", length_of_lis([10, 9, 2, 5, 3, 7, 101, 18]))
````

````python
def coin_change(coins, amount):
    if not isinstance(amount, int) or amount < 0 or any(not isinstance(c, int) or c <= 0 for c in coins):
        raise ValueError("amount must be nonnegative and coins positive integers")
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0

    for coin in coins:
        for i in range(coin, amount + 1):
            dp[i] = min(dp[i], dp[i - coin] + 1)

    return dp[amount] if dp[amount] != float('inf') else -1

print("Minimum coins:", coin_change([1, 2, 5], 11))
````

````python
def max_subarray_sum(nums):
    if not nums:
        raise ValueError("maximum subarray needs at least one number")
    current_sum = max_sum = nums[0]
    
    for num in nums[1:]:
        current_sum = max(num, current_sum + num)
        max_sum = max(max_sum, current_sum)

    return max_sum

print("Maximum subarray sum:", max_subarray_sum([-2, 1, -3, 4, -1, 2, 1, -5, 4]))
````

````python
class DisjointSet:
    def __init__(self, n):
        if not isinstance(n, int) or n < 0:
            raise ValueError("n must be a nonnegative integer")
        self.parent = list(range(n))
        self.rank = [0] * n

    def find_set(self, x):
        if not 0 <= x < len(self.parent):
            raise IndexError("vertex index out of range")
        if self.parent[x] != x:
            self.parent[x] = self.find_set(self.parent[x])  # Path compression
        return self.parent[x]

    def union_sets(self, x, y):
        root_x = self.find_set(x)
        root_y = self.find_set(y)

        if root_x != root_y:
            if self.rank[root_x] < self.rank[root_y]:
                self.parent[root_x] = root_y
            elif self.rank[root_x] > self.rank[root_y]:
                self.parent[root_y] = root_x
            else:
                self.parent[root_x] = root_y
                self.rank[root_y] += 1

sets = DisjointSet(5)
sets.union_sets(0, 1); sets.union_sets(1, 2)
print("Connected 0/2:", sets.find_set(0) == sets.find_set(2))
print("Connected 0/4:", sets.find_set(0) == sets.find_set(4))
````

````python
def has_cycle(graph):
    vertices = {vertex for edge in graph for vertex in edge}
    indices = {vertex: i for i, vertex in enumerate(vertices)}
    dsu = DisjointSet(len(vertices))
    for node1, node2 in graph:
        u, v = indices[node1], indices[node2]
        if dsu.find_set(u) == dsu.find_set(v):
            return True
        dsu.union_sets(u, v)
    return False

print("Cycle:", has_cycle([(10, 20), (20, 30), (30, 10)]))
print("Tree:", has_cycle([(10, 20), (20, 30)]))
````

````python
def kruskal(graph):
    vertices = {vertex for u, v, _ in graph for vertex in (u, v)}
    indices = {vertex: i for i, vertex in enumerate(vertices)}
    dsu = DisjointSet(len(vertices))
    minimum_spanning_tree = []
    for node1, node2, weight in sorted(graph, key=lambda edge: edge[2]):
        u, v = indices[node1], indices[node2]
        if dsu.find_set(u) != dsu.find_set(v):
            minimum_spanning_tree.append((node1, node2, weight))
            dsu.union_sets(u, v)
    return minimum_spanning_tree

edges = [(10, 20, 3), (20, 30, 1), (10, 30, 2)]
forest = kruskal(edges)
print("Minimum spanning tree:", forest)
print("Total weight:", sum(edge[2] for edge in forest))
````

def longest_word(words):
    trie = Trie()
    for word in sorted(words):
        trie.insert(word)
    
    longest = ""
    for word in words:

def length_of_lis(nums):
    if not nums:
        return 0

    dp = [1] * len(nums)

    for i in range(1, len(nums)):
        for j in range(i):
            if nums[i] > nums[j]:
                dp[i] = max(dp[i], dp[j] + 1)

    return max(dp)
```

def coin_change(coins, amount):
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0

    for coin in coins:
        for i in range(coin, amount + 1):
            dp[i] = min(dp[i], dp[i - coin] + 1)

    return dp[amount] if dp[amount] != float('inf') else -1
```

class DisjointSet:
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n

    def find_set(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find_set(self.parent[x])  # Path compression
        return self.parent[x]

    def union_sets(self, x, y):
        root_x = self.find_set(x)
        root_y = self.find_set(y)

        if root_x != root_y:
            if self.rank[root_x] < self.rank[root_y]:
                self.parent[root_x] = root_y
            elif self.rank[root_x] > self.rank[root_y]:
                self.parent[root_y] = root_x
            else:
                self.parent[root_x] = root_y
                self.rank[root_y] += 1
```

def has_cycle(graph):
    dsu = DisjointSet(len(graph))

    for edge in graph:
        node1, node2 = edge

        if dsu.find_set(node1) == dsu.find_set(node2):
            return True

        dsu.union_sets(node1, node2)

    return False
```
