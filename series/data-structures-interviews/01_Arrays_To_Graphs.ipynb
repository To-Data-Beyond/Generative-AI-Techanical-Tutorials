````python
def max_subarray_sum(nums):
    if not nums:
        raise ValueError("maximum subarray needs at least one number")
    max_sum = current_sum = nums[0]
    for num in nums[1:]:
        current_sum = max(num, current_sum + num)
        max_sum = max(max_sum, current_sum)
    return max_sum

print("Maximum subarray sum:", max_subarray_sum([-2, 1, -3, 4, -1, 2, 1, -5, 4]))
````

````python
def rotate_array(nums, k):
    if not nums:
        return
    n = len(nums)
    k = k % n
    nums[:] = nums[-k:] + nums[:-k]

values = [1, 2, 3, 4, 5]
rotate_array(values, 2)
print("Rotated array:", values)
````

````python
class MyArray:
    def __init__(self):
        self.length = 0
        self.data = {}

    def get(self, index):
        return self.data[index]

    def push(self, item):
        self.data[self.length] = item
        self.length += 1

    def pop(self):
        if self.length == 0:
            return None
        popped_item = self.data[self.length - 1]
        del self.data[self.length - 1]
        self.length -= 1
        return popped_item

array = MyArray()
array.push(10); array.push(20)
print("Array get / pop / length:", array.get(0), array.pop(), array.length)
````

````python
def single_number(nums):
    result = 0
    for num in nums:
        result ^= num
    return result

print("Single number:", single_number([4, 1, 2, 1, 2]))
````

````python
class ListNode:
    def __init__(self, value=0, next=None):
        self.value = value
        self.val = value
        self.next = next

def make_list(values):
    head = None
    for value in reversed(values):
        head = ListNode(value, head)
    return head

def list_values(head):
    result = []
    while head:
        result.append(head.value)
        head = head.next
    return result
print("Linked-list node and construction helpers ready.")

def reverse_linked_list(head):
    prev, current = None, head
    while current:
        next_node = current.next
        current.next = prev
        prev = current
        current = next_node
    return prev

head = make_list([1, 2, 3])
print("Reversed linked list:", list_values(reverse_linked_list(head)))
````

````python
def find_middle(head):
    if head is None:
        return None
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    return slow.value

print("Middle value:", find_middle(make_list([1, 2, 3, 4])))
````

````python
def merge_sorted_arrays(arr1, arr2):
    result = []
    i = j = 0
    while i < len(arr1) and j < len(arr2):
        if arr1[i] < arr2[j]:
            result.append(arr1[i])
            i += 1
        else:
            result.append(arr2[j])
            j += 1
    result.extend(arr1[i:])
    result.extend(arr2[j:])
    return result

print("Merged arrays:", merge_sorted_arrays([1, 3, 5], [2, 4, 6]))
````

````python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if not self.is_empty():
            return self.items.pop()
        else:
            return None

    def is_empty(self):
        return len(self.items) == 0

stack = Stack()
stack.push(1); stack.push(2)
print("Stack pop:", stack.pop())
````

````python
class MinStack:
    def __init__(self):
        self.stack = []
        self.min_stack = []

    def push(self, value):
        self.stack.append(value)
        if not self.min_stack or value <= self.min_stack[-1]:
            self.min_stack.append(value)

    def pop(self):
        if self.stack:
            popped_value = self.stack.pop()
            if popped_value == self.min_stack[-1]:
                self.min_stack.pop()
            return popped_value
        else:
            return None

    def get_min(self):
        return self.min_stack[-1] if self.min_stack else None

min_stack = MinStack()
min_stack.push(3); min_stack.push(1)
print("Minimum:", min_stack.get_min())
````

````python
def is_balanced_parentheses(s):
    stack = []
    brackets = {'(': ')', '[': ']', '{': '}'}

    for char in s:
        if char in brackets.keys():
            stack.append(char)
        elif char in brackets.values():
            if not stack or brackets[stack.pop()] != char:
                return False

    return len(stack) == 0

print("Balanced:", is_balanced_parentheses("{[()]}"))
````

````python
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, item):
        self.items.append(item)

    def dequeue(self):
        if not self.is_empty():
            return self.items.pop(0)
        else:
            return None

    def is_empty(self):
        return len(self.items) == 0

queue = Queue()
queue.enqueue(1); queue.enqueue(2)
print("Queue dequeue:", queue.dequeue())
````

````python
class MyCircularQueue:
    def __init__(self, k):
        if not isinstance(k, int) or k <= 0:
            raise ValueError("capacity must be a positive integer")
        self.capacity = k
        self.queue = [None] * k
        self._front = self._rear = -1

    def enqueue(self, value):
        if self.is_empty():
            self._front = self._rear = 0
        elif not self.is_full():
            self._rear = (self._rear + 1) % self.capacity
        else:
            return False
        self.queue[self._rear] = value
        return True

    def dequeue(self):
        if not self.is_empty():
            if self._front == self._rear:
                self._front = self._rear = -1
            else:
                self._front = (self._front + 1) % self.capacity
            return True
        else:
            return False

    def front(self):
        return self.queue[self._front] if not self.is_empty() else -1

    def rear(self):
        return self.queue[self._rear] if not self.is_empty() else -1

    def is_empty(self):
        return self._front == -1

    def is_full(self):
        return (self._rear + 1) % self.capacity == self._front

circular = MyCircularQueue(3)
circular.enqueue(1); circular.enqueue(2)
print("Front / rear / full:", circular.front(), circular.rear(), circular.is_full())
````

````python
from queue import Queue as StandardQueue

def reverse_k_elements(queue, k):
    if not isinstance(k, int) or not 0 <= k <= queue.qsize():
        raise ValueError("k must be between zero and the queue size")
    stack = []
    for _ in range(k):
        stack.append(queue.get())

    while stack:
        queue.put(stack.pop())

    for _ in range(queue.qsize() - k):
        queue.put(queue.get())

my_queue = StandardQueue()
for value in range(1, 6): my_queue.put(value)
reverse_k_elements(my_queue, 3)
print("Queue after reversal:", [my_queue.get_nowait() for _ in range(my_queue.qsize())])
````

````python
def inorder_traversal(root):
    result = []
    def visit(node):
        if node:
            visit(node.left)
            result.append(node.value)
            visit(node.right)
    visit(root)
    return result

def preorder_traversal(root):
    result = []
    def visit(node):
        if node:
            result.append(node.value)
            visit(node.left)
            visit(node.right)
    visit(root)
    return result

def postorder_traversal(root):
    result = []
    def visit(node):
        if node:
            visit(node.left)
            visit(node.right)
            result.append(node.value)
    visit(root)
    return result

print("Inorder:", inorder_traversal(root))
print("Preorder:", preorder_traversal(root))
print("Postorder:", postorder_traversal(root))
````

````python
def is_balanced(root):
    def height(node):
        if not node:
            return 0
        left_height = height(node.left)
        right_height = height(node.right)
        if left_height == -1 or right_height == -1 or abs(left_height - right_height) > 1:
            return -1
        return max(left_height, right_height) + 1

    return height(root) != -1

print("Tree balanced:", is_balanced(root))
````

````python
def lowest_common_ancestor(root, p, q):
    if not root:
        return None

    if root.value > p.value and root.value > q.value:
        return lowest_common_ancestor(root.left, p, q)
    elif root.value < p.value and root.value < q.value:
        return lowest_common_ancestor(root.right, p, q)
    else:
        return root

print("Lowest common ancestor:", lowest_common_ancestor(root, root.left.left, root.left.right).value)
````

````python
import heapq
from itertools import count

def dijkstra(graph, start):
    vertices = set(graph) | {v for neighbors in graph.values() for v in neighbors} | {start}
    if any(weight < 0 for neighbors in graph.values() for weight in neighbors.values()):
        raise ValueError("Dijkstra requires nonnegative weights")
    distances = {vertex: float('infinity') for vertex in vertices}
    distances[start] = 0
    order = count()
    priority_queue = [(0, next(order), start)]
    while priority_queue:
        current_distance, _, current_vertex = heapq.heappop(priority_queue)
        if current_distance > distances[current_vertex]:
            continue
        for neighbor, weight in graph.get(current_vertex, {}).items():
            distance = current_distance + weight
            if distance < distances[neighbor]:
                distances[neighbor] = distance
                heapq.heappush(priority_queue, (distance, next(order), neighbor))
    return distances

weighted = {1: {2: 1, 3: 4}, 2: {3: 2}, 3: {}}
print("Shortest distances:", dijkstra(weighted, 1))
````

import heapq

def dijkstra(graph, start):
    distances = {vertex: float('infinity') for vertex in graph}
    distances[start] = 0
    priority_queue = [(0, start)]

    while priority_queue:
        current_distance, current_vertex = heapq.heappop(priority_queue)

        if current_distance > distances[current_vertex]:
            continue

        for neighbor, weight in graph[current_vertex].items():
            distance = current_distance + weight
            if distance < distances[neighbor]:
                distances[neighbor] = distance
                heapq.heappush(priority_queue, (distance, neighbor))

# Example usage:
# my_weighted_graph = {1: {2: 1, 3: 4}, 2: {3: 2}, 3: {}}
# dijkstra(my_weighted_graph, 1)
