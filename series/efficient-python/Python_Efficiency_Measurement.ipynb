## 1. Compare doubling styles


```python
numbers = [1, 2, 3, 4, 5]

# Non-Pythonic
doubled_numbers = []

for i in range(len(numbers)):
  doubled_numbers.append(numbers[i] * 2)

# Pythonic
doubled_numbers = [x * 2 for x in numbers]
print(doubled_numbers)
```

## 2. range examples


```python
# range(start,stop)
nums = range(0,11)
nums_list = list(nums)
print(nums_list)

# range(stop)
nums = range(11)
nums_list = list(nums)
print(nums_list)

# Using range() with a step value

even_nums = range(2, 11, 2)
even_nums_list = list(even_nums)
print(even_nums_list)
```

## 9. The intentional list TypeError


```python
# Python lists do not support element-wise exponentiation.
nums = [-2, -1, 0, 1, 2]
try:
    nums ** 2
except TypeError as error:
    print(f"{type(error).__name__}: {error}")
```

## 10. Measure three squaring approaches


```python
import time

# define numerical list 
nums = range(0,1000)
nums = list(nums)

# For loop (inefficient option)
# get the start time
st = time.perf_counter()

sqrd_nums = []
for num in nums:
    sqrd_nums.append(num ** 2)
#print(sqrd_nums)

# get the end time
et = time.perf_counter()

# get the execution time
elapsed_time = et - st
print('Execution time using for loops over list:', elapsed_time, 'seconds')


# List comprehension (better option but not best)
# get the start time
st = time.perf_counter()

sqrd_nums = [num ** 2 for num in nums]
#print(sqrd_nums)

# get the end time
et = time.perf_counter()
elapsed_time = et - st
print('Execution time using list comprehension:', elapsed_time, 'seconds')


# using numpy array broadcasting

# define the numpy array 
nums_np = np.arange(0,1000)
# get the start time
st = time.perf_counter()

nums_np ** 2

# get the end time
et = time.perf_counter()

# get the execution time
elapsed_time = et - st
print('Execution time using numpy array broadcasting:', elapsed_time, 'seconds')
```

## 11. Two-dimensional indexing


```python
#2-D list
nums2 = [ [1, 2, 3],
        [4, 5, 6] ]


# 2-D array
nums2_np = np.array(nums2)

# printing the second item of the first row 
print(nums2[0][1])
print(nums2_np[0,1])

# printing the first row values 

print([row[0] for row in nums2])
print(nums2_np[:,0])
```

## 12. Boolean indexing


```python
nums = [-2, -1, 0, 1, 2]
nums_np = np.array(nums)

# Boolean indexing
print(nums_np[nums_np > 0])

# No boolean indexing for lists
# For loop (inefficient option)


pos = []
for num in nums:
    if num > 0:
        pos.append(num)
print(pos)


# List comprehension (better option but not best)
pos = [num for num in nums if num > 0]
print(pos)
```

## 19. Convert height and weight units


```python
def convert_units(names, heights, weights):
    new_hts = [ht * 0.39370 for ht in heights]
    new_wts = [wt * 2.20462 for wt in weights]
    data = {}
    for i,name in enumerate(names):
        data[name] = (new_hts[i], new_wts[i])
    return data
print(convert_units(names, hts, wts))

notebook_convert_units = convert_units
```

## Temporary source module for memory profiling


```python
# Use a temporary module: memory_profiler requires physical source lines.
from pathlib import Path
from tempfile import TemporaryDirectory
import importlib.util
import sys

profile_module_dir = TemporaryDirectory(prefix="tdb-profiling-", dir=".")
module_path = Path(profile_module_dir.name) / "utils_funcs.py"
module_source = """def convert_units(names, heights, weights):
    new_hts = [ht * 0.39370 for ht in heights]
    new_wts = [wt * 2.20462 for wt in weights]
    data = {}
    for i, name in enumerate(names):
        data[name] = (new_hts[i], new_wts[i])
    return data
"""
module_path.write_text(module_source, encoding="utf-8")
spec = importlib.util.spec_from_file_location("utils_funcs", module_path)
module = importlib.util.module_from_spec(spec)
sys.modules["utils_funcs"] = module
spec.loader.exec_module(module)
print("Created temporary utils_funcs.py for line-by-line memory profiling.")
```

## Regression checks

Verify example values and matching algorithms, timing-result structure, accurate conversions, unchanged inputs, file-backed source and real profiler hits. No absolute timings or rankings are asserted.


```python
import sys
from line_profiler import LineProfiler

assert doubled_numbers == [2, 4, 6, 8, 10]
assert even_nums_list == [2, 4, 6, 8, 10]
assert indexed_letters_list == [(0, 'a'), (1, 'b'), (2, 'c'), (3, 'd')]
assert indexed_letters2_list == [(5, 'a'), (6, 'b'), (7, 'c'), (8, 'd')]
assert nums_np_ints.dtype.kind in 'iu'
assert nums_np_floats.dtype.kind == 'f'
np.testing.assert_allclose(nums_np_floats, [1, 2.5, 3])
assert nums2[0][1] == nums2_np[0, 1] == 2
assert [row[0] for row in nums2] == nums2_np[:, 0].tolist() == [1, 4]
assert pos == [1, 2]
assert sqrd_nums == (np.arange(1000) ** 2).tolist()
try:
    [1, 2] ** 2
except TypeError:
    pass
else:
    raise AssertionError('Lists do not implement element-wise exponentiation')

assert len(times.timings) == 7
assert times.loops == 1000 and 0 < times.best <= times.worst
assert elapsed_time >= 0
before_heights, before_weights = hts.copy(), wts.copy()
expected = {'Ahmed': (74.0156, 209.4389), 'Mohammed': (75.1967, 220.462), 'Youssef': (72.8345, 165.3465)}
result = convert_units(names, hts, wts)
assert set(result) == set(expected)
for name in names:
    np.testing.assert_allclose(result[name], expected[name])
    np.testing.assert_allclose(notebook_convert_units(names, hts, wts)[name], result[name])
np.testing.assert_array_equal(before_heights, hts)
np.testing.assert_array_equal(before_weights, wts)
assert convert_units([], np.array([]), np.array([])) == {}

profiler = LineProfiler(convert_units)
assert profiler.runcall(convert_units, names, hts, wts) == result
stats = profiler.get_stats()
assert stats.unit > 0 and stats.timings
assert any(hits >= 3 for rows in stats.timings.values() for _, hits, _ in rows)
assert module_path.is_file() and 'def convert_units' in module_path.read_text()
assert sys.getsizeof(nums_list) > 0 and sys.getsizeof(nums_np) > 0
assert version('line_profiler') == '5.1.0'
assert version('memory_profiler') == '0.61.0'
print('All value, timing-result, conversion, input-preservation and profiler checks passed.')
print('Conversions:', {name: tuple(float(value) for value in values) for name, values in result.items()})
print('Profiled functions:', len(stats.timings))
profile_module_dir.cleanup()
```
