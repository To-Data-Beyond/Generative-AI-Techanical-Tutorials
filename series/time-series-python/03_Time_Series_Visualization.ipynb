````python
# Decomposing multiple time series with Python dictionaries
# Import the statsmodel library
import statsmodels.api as sm
# Initialize a dictionary
my_dict = {}
# Extract the names of the time series
ts_names = jobs.columns
print(ts_names)
for figure_number in plt.get_fignums():
    display(plt.figure(figure_number))
plt.close('all')
print('Example 42 completed.')
````

````python
# Run time series decomposition
for ts in ts_names:
    ts_decomposition = sm.tsa.seasonal_decompose(jobs[ts],period=12)
    my_dict[ts] = ts_decomposition
    
# Initialize a new dictionnary for each component
my_dict_trend = {}
my_dict_seasonal = {}
my_dict_resid = {}
# Extract the trend component
for ts in ts_names:
    my_dict_trend[ts] = my_dict[ts].trend
    my_dict_seasonal[ts] = my_dict[ts].seasonal
    my_dict_resid[ts] = my_dict[ts].resid

# Convert to a DataFrame
trend_df = pd.DataFrame.from_dict(my_dict_trend)
seasonal_df = pd.DataFrame.from_dict(my_dict_seasonal)
resid_df = pd.DataFrame.from_dict(my_dict_resid)


trend_df.plot(subplots=True,layout=(4,4),figsize=(20,14));plt.show()
seasonal_df.plot(subplots=True,layout=(4,4),figsize=(20,14));plt.show()
resid_df.plot(subplots=True,layout=(4,4),figsize=(20,14));plt.show()
for figure_number in plt.get_fignums():
    display(plt.figure(figure_number))
plt.close('all')
print('Example 43 completed.')
````

````python
from numpy.testing import assert_allclose
from scipy.stats import spearmanr
from scipy.cluster.hierarchy import linkage,leaves_list
# Independently reconstruct the centered moving-average trend and additive components.
for observed,result,period in [(co2_levels['co2'],co2_decomp,52),(airline.iloc[:,0],decomposition,12)]+[(jobs[c],my_dict[c],12)for c in jobs]:
    weights=np.r_[.5,np.ones(period-1),.5]/period
    trend=np.convolve(observed.to_numpy(),weights,mode='valid')
    half=period//2
    assert_allclose(result.trend.iloc[half:-half],trend,atol=1e-10)
    valid=result.trend.notna()
    assert_allclose((result.trend+result.seasonal+result.resid)[valid],observed[valid],atol=1e-10)
    phase_means=np.array([np.nanmean((observed-result.trend).to_numpy()[k::period])for k in range(period)])
    phase_means-=phase_means.mean()
    assert_allclose(result.seasonal.to_numpy(),np.resize(phase_means,len(observed)),atol=1e-10)
for a in meat.columns:
    for b in meat.columns:
        pair=meat[[a,b]].dropna().to_numpy()
        expected=np.corrcoef(pair[:,0],pair[:,1])[0,1]
        assert_allclose(corr_mat.loc[a,b],expected,atol=1e-12)
assert jobs.mean().idxmax()=='Agriculture'and jobs.std().idxmax()=='Construction'
assert jobs.shape==(122,16)and len(airline)==144
assert not co2_levels.isna().any().any()
assert len(my_dict)==len(jobs.columns)==16
expected_leaves=leaves_list(linkage(trend_corr.to_numpy(),method='average',metric='euclidean'))
assert list(fig.dendrogram_row.reordered_ind)==list(expected_leaves)
metrics={'co2_rows':len(co2_levels),'airline_rows':len(airline),'jobs_shape':jobs.shape,'highest_mean':jobs.mean().idxmax(),'highest_std':jobs.std().idxmax(),'decomposed_series':18}
print('All 18 full decompositions match independent convolution/phase means and reconstruct every interior observation.')
print('Every meat correlation, job summary, full row counts and hierarchical cluster order passed.')
print('METRICS:',metrics)
os.chdir(previous_cwd)
workspace.cleanup()
print('Owned temporary workspace cleaned up.')
````
