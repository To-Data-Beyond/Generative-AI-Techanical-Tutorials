````python
!pip install -q -U google-genai
````

Then, initialize the client and start generating content:


````python
from google import genai
import os
from google.colab import userdata

GEMINI_API_KEY = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=GEMINI_API_KEY)

response = client.models.generate_content(
    model="gemini-2.5-pro-exp-03-25",
    contents="Explain the role of encoder in the Transformers"
)

print(response.text)
````

**Expected output**

```text
Okay, let’s break down the role of the **Encoder** in the Transformer architecture.

Think of the original Transformer (like the one used for machine translation) as having two main parts: an **Encoder** and a **Decoder**.

**The Primary Role of the Encoder:**

The encoder’s main job is to **process the entire input sequence** (e.g., a sentence in the source language) and **transform it into a sequence of rich, contextualized representations (embeddings)**. These representations capture the meaning of each word *in the context of all other words* in the input sequence.

**Think of it like this:**

Imagine you’re reading a sentence. You don’t just understand each word in isolation. You understand “bank” differently in “I sat on the river bank” versus “I went to the savings bank.” The encoder does something similar for the machine. It reads the whole input sentence and figures out the nuanced meaning of each word based on its surroundings.

**How it Achieves This (Key Components and Steps):**

1. **Input Embeddings:** The input sequence (words or sub-word tokens) is first converted into numerical vectors (embeddings). These initial embeddings represent the words but lack contextual understanding.

2. **Positional Encoding:** Since Transformers don’t process words sequentially like RNNs (they look at all words at once via attention), they need a way to know the *order* of the words. Positional encodings are vectors added to the input embeddings to give the model information about the position of each token in the sequence.

3. **Encoder Blocks (Stacked N times):** The core of the encoder consists of multiple identical blocks stacked on top of each other (e.g., 6 blocks in the original paper). Each block performs two main operations:
   * **Multi-Head Self-Attention:**
     * **Self-Attention:** This is the key mechanism. For each word in the input sequence, self-attention calculates an “attention score” determining how relevant every *other* word (including itself) in the *same input sequence* is to understanding this particular word.
     * **Weighted Sum:** It then creates a new representation for the word by taking a weighted sum of the representations of all words in the sequence, where the weights are the attention scores. Words deemed more relevant get higher weights. This directly injects context.
     * **Multi-Head:** Instead of doing this once, the model does it multiple times in parallel (multiple “heads”). Each head can potentially learn different types of relationships or focus on different aspects of the context (e.g., syntactic dependencies, semantic similarities). The results from the heads are combined.
   * **Position-wise Feed-Forward Network:**
     * After the attention mechanism, the resulting representation for each position is passed *independently* through a simple fully connected feed-forward network (typically two linear layers with a ReLU activation in between).
     * This network further processes the representations, adding more computational capacity and non-linearity, allowing the model to learn more complex transformations of the context-aware embeddings.

4. **Residual Connections and Layer Normalization:** Around each of the two main sub-layers (Self-Attention and Feed-Forward Network) within an encoder block, there are:
   * **Residual Connections:** The input to the sub-layer is added to the output of the sub-layer (x + Sublayer(x)). This helps with training deeper networks by allowing gradients to flow more easily and preventing information loss.
   * **Layer Normalization:** Applied after the addition. It stabilizes the activations and speeds up training.

**Output of the Encoder:**

The final output of the encoder stack is a sequence of vectors (embeddings), one for each input token. Crucially, each vector now contains not just information about the original token but also rich contextual information derived from the entire input sequence via the self-attention mechanism.

**What happens to the Encoder’s output?**

* **In Encoder-Decoder Models (e.g., Translation):** This sequence of contextualized embeddings is passed to the **Decoder**. Specifically, it’s used in the “Encoder-Decoder Attention” (or “Cross-Attention”) layers within each decoder block. The decoder uses these representations to understand the meaning of the source sentence while generating the target sentence.
* **In Encoder-Only Models (e.g., BERT):** Models like BERT use *only* the encoder stack. The final hidden states (the output embeddings) from the encoder are then used directly as input features for downstream tasks like text classification, named entity recognition, question answering, etc., usually by adding a small task-specific layer on top.

In summary, the encoder’s role is to **understand the input sequence thoroughly** by creating representations where each element is aware of the context provided by all other elements, primarily using the powerful mechanism of self-attention.
```

## 2. OpenRouter API

![OpenRouter banner describing a unified interface for language models](https://todatabeyond.com/articles/gemini-api-access/openrouter-overview.png)

*OpenRouter offers a unified interface for comparing and accessing language models from multiple providers.*

OpenRouter is a flexible API gateway that connects you to a variety of AI models, especially large language models from top providers like Anthropic, Google, and Meta.

It offers a unified interface, making it easy for developers to interact with different models using just a single client and an API key—no need for separate setups.

[Open OpenRouter](https://openrouter.ai/)

### 2.1. Key Features

- Model Routing: Effortlessly switch between different AI models.
- Provider Routing: Access models across multiple providers without extra configuration.
- Multimodal Support: Upload images, audio, text, and files, and generate either text or images in response.
- Flexible Pricing: Choose from a mix of free and paid model options.

### 2.2. How to Get Started

Go to the OpenRouter website and sign up for an account.

![OpenRouter landing page with Gemini 2.5 Pro Preview highlighted among featured models](https://todatabeyond.com/articles/gemini-api-access/openrouter-dashboard.png)

*The OpenRouter interface lists Gemini 2.5 Pro Preview alongside models from other providers.*

Generate an API key and set it as an environment variable on your system.

![OpenRouter model catalog filtered for Google Gemini 2.5 with the account menu open](https://todatabeyond.com/articles/gemini-api-access/openrouter-api-key.png)

*After signing in, use the account menu’s Keys option to create an OpenRouter API key.*

You can interact with OpenRouter using the Python requests package, the OpenAI client, or a simple CURL command. To use the OpenAI Python client, first install it:


````python
!pip install openai
````

Then, initialize the client and generate a response:


````python
from openai import OpenAI
import os

client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.environ["OPENROUTER"],
)

completion = client.chat.completions.create(
    model="google/gemini-2.5-pro-exp-03-25:free",
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "type": "text",
                    "text": "Write a poem about the image."
                },
                {
                    "type": "image_url",
                    "image_url": {
                        "url": "https://unsplash.com/s/photos/palestine"
                    }
                }
            ]
        }
    ]
)

print(completion.choices[0].message.content)
````

## 3. Requesty

![Requesty unified LLM platform banner highlighting routing, cost management, analytics, and logging](https://todatabeyond.com/articles/gemini-api-access/requesty-overview.png)

*Requesty combines smart model routing with cost management, analytics, and request logging.*

Requesty is a developer platform built for easy integration with large language models like Gemini 2.5 Pro. It offers smart LLM routing, dynamically selecting the best model to optimize performance based on each request.

[Open Requesty](https://www.requesty.ai/)

### 3.1. Key Features

- Developer-Friendly: Streamlines integration with Gemini 2.5 Pro.
- Intelligent Routing: Automatically picks the most suitable model for the task.
- Interactive Dashboard: Track and manage your API activity in real time.
- Production-Ready: Built to support scalable, real-world applications.

### 3.2. Getting Started

1. Visit the Requesty website and sign up for an account.
2. During account setup, you’ll be prompted to generate your first API key. Save it as an environment variable for later use.

To interact with Requesty, use the OpenAI Python client. First, install the OpenAI client if you haven’t already:


````python
!pip install openai
````

Then, initialize the client and send your first request:


````python
from openai import OpenAI
import os

REQUESTY_API_KEY = userdata.get("REQUESTY_API_KEY")

client = OpenAI(
    base_url="https://router.requesty.ai/v1",
    api_key=REQUESTY_API_KEY,
)

completion = client.chat.completions.create(
    model="google/gemini-2.5-pro-exp-03-25",
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "type": "text",
                    "text": "How AI will impact software engineerg?."
                }
            ]
        }
    ]
)

print(completion.choices[0].message.content)
````
