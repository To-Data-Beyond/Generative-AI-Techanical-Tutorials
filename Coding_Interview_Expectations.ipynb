````python
def find_median_part1(nums):
    if not nums:
        raise ValueError("median needs at least one number")
    sorted_nums = sorted(nums)
    n = len(sorted_nums)
    if n % 2 == 0:
        return (sorted_nums[n // 2 - 1] + sorted_nums[n // 2]) / 2
    return sorted_nums[n // 2]

print("Odd median:", find_median_part1([7, 1, 3]))
print("Even median:", find_median_part1([7, 1, 3, 9]))
````

````python
import random

def estimate_pi(num_samples):
    if not isinstance(num_samples, int) or num_samples <= 0:
        raise ValueError("num_samples must be a positive integer")
    inside_circle = 0
    for _ in range(num_samples):
        x, y = random.uniform(-1, 1), random.uniform(-1, 1)
        if x**2 + y**2 <= 1:
            inside_circle += 1
    return (inside_circle / num_samples) * 4

random.seed(42)
pi_estimate = estimate_pi(100_000)
print("Estimated pi:", pi_estimate)
````

````python
from math import isqrt

def generate_primes_up_to_n(n):
    if n < 2:
        return []
    is_prime = [True] * (n + 1)
    is_prime[0] = is_prime[1] = False
    for num in range(2, isqrt(n) + 1):
        if is_prime[num]:
            for multiple in range(num * num, n + 1, num):
                is_prime[multiple] = False
    return [num for num in range(2, n + 1) if is_prime[num]]

print("Primes up to 30:", generate_primes_up_to_n(30))
````

````python
import numpy as np

def simulate_multinomial_distribution(probabilities, num_trials=1, seed=42):
    probabilities = np.asarray(probabilities, dtype=float)
    if (probabilities.ndim != 1 or len(probabilities) == 0
            or not np.all(np.isfinite(probabilities))
            or np.any(probabilities < 0)
            or not np.isclose(probabilities.sum(), 1.0)):
        raise ValueError("probabilities must be finite, nonnegative and sum to 1")
    if not isinstance(num_trials, (int, np.integer)) or num_trials < 0:
        raise ValueError("num_trials must be a nonnegative integer")
    probabilities = probabilities / probabilities.sum()
    cumulative = np.cumsum(probabilities)
    cumulative[-1] = 1.0
    uniforms = np.random.default_rng(seed).uniform(0.0, 1.0, num_trials)
    categories = np.searchsorted(cumulative, uniforms, side="right")
    return np.bincount(categories, minlength=len(probabilities))

counts = simulate_multinomial_distribution([0.2, 0.3, 0.5], 1000)
print("Multinomial counts:", counts.tolist())
print("Total trials:", int(counts.sum()))
````

````python
import math
for n in range(1, 7):
    for values in product(range(-2, 3), repeat=n):
        assert find_median_part1(values) == median(values)
for n in range(-5, 301):
    expected = [k for k in range(2, n + 1) if all(k % d for d in range(2, isqrt(k) + 1))]
    assert generate_primes_up_to_n(n) == expected
assert abs(pi_estimate - math.pi) < 0.03
state = random.getstate()
random.seed(7); reference = sum(random.uniform(-1, 1)**2 + random.uniform(-1, 1)**2 <= 1 for _ in range(200)) * 4 / 200
random.seed(7); assert estimate_pi(200) == reference
random.setstate(state)
for probabilities in ([1.0], [0.0, 1.0], [0.2, 0.3, 0.5], [0.25] * 4):
    draws = np.random.default_rng(19).uniform(size=1000)
    p = np.asarray(probabilities); p = p / p.sum()
    expected = [0] * len(p)
    for draw in draws:
        running = 0.0
        for j, probability in enumerate(p):
            running += probability
            if draw < running or j == len(p) - 1:
                expected[j] += 1; break
    actual = simulate_multinomial_distribution(probabilities, 1000, 19)
    assert actual.tolist() == expected
    assert actual.sum() == 1000 and np.all(actual >= 0)
assert simulate_multinomial_distribution([0.2, 0.3, 0.5], 0).tolist() == [0, 0, 0]
empirical = simulate_multinomial_distribution([0.2, 0.3, 0.5], 200_000, 91) / 200_000
assert np.max(np.abs(empirical - [0.2, 0.3, 0.5])) < 0.005
for bad in ([], [-0.1, 1.1], [0.2, 0.2], [float('nan'), 1]):
    try: simulate_multinomial_distribution(bad)
    except ValueError: pass
    else: raise AssertionError('invalid probabilities accepted')
for function, args in [(find_median_part1, ([],)), (estimate_pi, (0,)), (simulate_multinomial_distribution, ([1.0], -1))]:
    try: function(*args)
    except ValueError: pass
    else: raise AssertionError('invalid input accepted')
print("19,530 exhaustive median cases; primes through 300; seeded Monte Carlo and inverse-CDF multinomial references passed.")
print("Invalid inputs, zero trials, deterministic probabilities and 200,000-draw frequency checks passed.")
````

import numpy as np

def simulate_multinomial_distribution(probabilities):
    num_categories = len(probabilities)
    random_uniform = np.random.uniform(0, 1, num_categories - 1)
    sorted_uniform = np.sort(random_uniform)
    intervals = np.concatenate(([0], sorted_uniform, [1]))
    counts = np.diff(intervals) * probabilities.sum()
    return counts.astype(int)
