## Running the examples

Run each notebook from top to bottom in its own fresh environment using its setup instructions. First-run package/model/data downloads need internet. Notebooks are committed without saved outputs.

The first three LangChain companions run their local default paths without paid keys; scripted parser/wiring checks are not hosted inference. Their hosted appendices remain opt-in and were not executed in this batch. Chains, agents and evaluation use a locally configured OpenAI key and incur charges. Weather tools in the introductory agent loop return fixed examples; the later graph performs real web searches. No unrestricted Python execution tool is enabled.
